```python
import os
import json
import tokenize
from io import StringIO
from google.colab import drive

# Connect Google Drive
drive.mount("/content/drive")

# Notebook name
filename = "Day_002_2026-09-25_Lesson_1_2_Manual.ipynb"

# Find the notebook
archivo = None

for root, dirs, files in os.walk("/content/drive/MyDrive"):
    if filename in files:
        archivo = os.path.join(root, filename)
        break

# Check if the notebook was found
if archivo is None:
    print("ERROR: Notebook not found.")

else:
    print(f"Notebook found: {archivo}")
    print("\nCOMMENTS + STRINGS + NUMBERS:\n")

    # Open the notebook
    with open(archivo, "r", encoding="utf-8") as file:
        notebook = json.load(file)

    # Read each Python code cell
    for cell in notebook["cells"]:

        if cell["cell_type"] == "code":

            code = "".join(cell["source"])

            # Break the Python code into tokens
            tokens = tokenize.generate_tokens(
                StringIO(code).readline
            )

            # Show comments, strings, and numbers
            for token in tokens:

                if token.type in (
                    tokenize.COMMENT,
                    tokenize.STRING,
                    tokenize.NUMBER
                ):
                    print(token.string)
```

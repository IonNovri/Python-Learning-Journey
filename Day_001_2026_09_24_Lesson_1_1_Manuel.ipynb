```python
# Raw data from client
raw_name = "  nordic senior living  "
raw_country = "norway"
residents = 45
contract_date = "2026-09-24"
raw_contact = "  Erik larsen "

# Clean data
clean_name = raw_name.strip().title()
clean_country = raw_country.strip().title()
clean_contact = raw_contact.strip().title()

# Report
line = "=" * 40

print(line)
print("NORDIC SENIOR LIVING - CLIENT REPORT")
print(line)
print(f"Client: {clean_name}")
print(f"Country: {clean_country}")
print(f"Residents: {residents}")
print(f"Contract date: {contract_date}")
print(f"Contact: {clean_contact}")
print(line)
```

```python
# Japanese client raw data
raw_name = "  sakura care home  "
raw_country = "japan"
raw_date_jp = "2026/09/24"

# Normalize
clean_name = raw_name.strip().title()
clean_country = raw_country.strip().title()
clean_date = raw_date_jp.replace("/", "-")

# Report
line = "=" * 40

print(line)
print("SAKURA CARE HOME - CLIENT REPORT")
print(line)
print(f"Client: {clean_name}")
print(f"Country: {clean_country}")
print(f"Contract date: {clean_date}")
print(line)
```

```python
# Raw sensor line
raw_line = "ROBOT_042_STATUS="

# Extract parts
parts = raw_line.split("=")
robot_id = parts[0]
status_value = parts[1]

# Detect missing
if status_value == "":
    status_value = "UNKNOWN"

# Display
print(f"Robot: {robot_id}")
print(f"Status: {status_value}")
```
